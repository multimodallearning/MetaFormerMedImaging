```python
from models.poolformer_classifier import FlexFormerClassifier

m_untrained = FlexFormerClassifier('imagewoof')
```

```python
import torch

with torch.no_grad():
    untrained_params = torch.cat([p.flatten() for p in m_untrained.parameters() if p.requires_grad])
    print(untrained_params.mean(), untrained_params.std())
    untrained_flex_params = torch.cat([p.flatten() for p in m_untrained.model.model.network[0][0].token_mixer.parameters() if p.requires_grad])
    print(untrained_flex_params.mean(), untrained_flex_params.std())
```
